```python
import numpy as np
```

### Scalar Arithmatic

```python
array = np.array([1,2,3])
```

```python
print(array + 1)  # -> This adds "1" to every single elements in the array
print(array - 2)  # -> This subtracts "2" to every single elements in the array
print(array * 3)  # -> This multiplies "3" to every single elements in the array
print(array / 4)  # -> This divides by "4" to every single elements in the array
```

### Verctorized math functions

```python
print(np.sqrt(array))
```

```python
# Rounding off the numbers within the array:
array1 = np.array([1.01, 2.30, 4.98])
print(np.round(array1))

print(np.floor(array1))
print(np.ceil(array1))
```

```python
# Pi value
print(np.pi)
```

#### Excercise with vectorized math function

```python
radii = np.array([1, 2, 3])
```

```python
# Area of a circle
print(np.pi * radii ** 2) # -> pi * r^2
```

#### Element-wise arithmetic

```python
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
```

```python
print(array1 + array2)
print(array1[0] * array2[0])
print(array1[2] + array2[1])
```

#### Comparision Operator

```python
print(array1 == array2)
print(array1 == 3)
```

```python
scores = np.array([91, 55, 100, 73, 82, 64])
scores [scores < 60] = 0
print(scores)
```
